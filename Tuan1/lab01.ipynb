BÀI 9

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)

print("\n===== YÊU CẦU 1 =====")

correct = torch.tensor([
    [1.0],
    [2.0],
    [3.0]
])

print("Hệ số học được:")
print(w)

print("\nHệ số đúng:")
print(correct)

print("\nSai số:")
print(w - correct)

print("\n===== YÊU CẦU 2 =====")

print("optimizer.zero_grad() : ")
print("mse.backward()       : ")
print("optimizer.step()     : ")
```

BÀI 10

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam(
    [A, B, C, D],
    lr=0.01
)
for _ in range(EPOCHS):

    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)

print("\n===== YÊU CẦU 1 =====")

print("A + B =", A + B)
print("C - D =", C - D)
print("A + C =", A + C)
print("B - D =", B - D)

print("\nSai số:")

print("PT1:", A + B - 9)
print("PT2:", C - D - 1)
print("PT3:", A + C - 8)
print("PT4:", B - D - 2)

print("\n===== YÊU CẦU 2 =====")

print("Nghiệm tổng quát:")
print("A = t")
print("B = 9 - t")
print("C = 8 - t")
print("D = 7 - t")
```

```python
import torch
import numpy as np

X = torch.arange(1, 25)

print("Tensor X:")
print(X)

print("\nReshape 4 x 6:")
print(X.reshape(4, 6))

print("\nReshape 2 x 12:")
print(X.reshape(2, 12))

print("\nReshape 2 x 3 x 4:")
print(X.reshape(2, 3, 4))

A = torch.randn(3, 4)
B = torch.randn(3, 4)

print("\nTensor A:")
print(A)

print("\nTensor B:")
print(B)

print("\nA + B:")
print(A + B)

print("\nA * B:")
print(A * B)

print("\nA / B:")
print(A / B)

print("\nTrung bình theo cột:")
print(torch.mean(A, dim=0))

print("\nĐộ lệch chuẩn theo cột:")
print(torch.std(A, dim=0))

x = torch.tensor(2.0, requires_grad=True)

y = 2 * x**2 + 5 * x + 3

y.backward()

print("\n===== AUTOGRAD =====")
print("x =", x)
print("y =", y)
print("dy/dx =", x.grad)

polynomial = np.poly1d([2, -3, 5])

N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)

x_train = torch.tensor(XX, dtype=torch.float32)
y_train = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

for epoch in range(1000):
    optimizer.zero_grad()

    y_pred = x_train @ w

    mse = torch.mean(torch.square(y_train - y_pred))

    mse.backward()

    optimizer.step()

print("\n===== KẾT QUẢ HUẤN LUYỆN =====")
print("Các hệ số tìm được:")
print(w)
```
